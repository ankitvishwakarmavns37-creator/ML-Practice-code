```python
##  t-SNE(Unsupervised (Mostly use for visualization 2D/3D)), LDA (Supervised(data, target(Lable)))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('iris')
```

```python
df.head(3)
```

```python
df.shape
```

```python
## Encoding column name(spacies)
df.species=df.species.map (lambda val: list(df.species.unique()).index(val))
```

```python
df.head(3)
```

```python
## Scalling
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler=MinMaxScaler()
```

```python
x=df.drop(columns=['species'])
y=df['species']
```

```python
x=scaler.fit_transform(x)
```

```python
x[:3]
```

### Using TSNE

```python
from sklearn.manifold import TSNE
```

```python
tsne=TSNE(n_components=2)
```

```python
tsne_x= tsne.fit_transform(x)
```

```python
tsne_x[:3]
```

```python
plt.figure(figsize=(8,3))
plt.scatter(tsne_x[:,0], tsne_x[:,1])
plt.xlabel('TSNE1')
plt.ylabel('TSNE2')
plt.show()


```

```python
plt.figure(figsize=(8,3))
plt.scatter(x[:,0], x[:,1],c=y)
plt.xlabel('PCA1')
plt.ylabel('PCA2')
plt.show()
```

### Using LDA(Linear Discriminant Analsis)

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
LDA=LinearDiscriminantAnalysis(n_components=2)
```

```python
lda_x =LDA.fit_transform(x,y) 
```

```python
plt.figure(figsize=(8,3))
plt.scatter(lda_x[:,0], lda_x[:,1] ,c=y)
plt.xlabel('LDA1')
plt.ylabel('LDA2')
plt.show()
```

### Using KMeans (Clustring/Grouping)

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=3)
```

```python
kmeans.fit(tsne_x)
```

```python
labels=kmeans.labels_
```

```python
## OR
```

```python
lables=kmeans.fit_predict(tsne_x)
```

```python
labels
```

```python
plt.figure(figsize=(8,3))
plt.scatter(tsne_x[:,0], tsne_x[:,1])
plt.xlabel('TSNE1')
plt.ylabel('TSNE2')
plt.show()
```

```python
plt.figure(figsize=(12,3))
plt.subplot(1,2,1)
plt.scatter(tsne_x[:,0], tsne_x[:,1],c=y)
plt.title("Original")
plt.subplot(1,2,2)
plt.scatter(tsne_x[:,0], tsne_x[:,1],c=labels)
plt.title("t-SNE")
plt.show()
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y,labels)*100
```
