### Standardization and Normalization

```python
data={
    "Empid": [ x for x in range(101,111)],
    "Empsalary": [1234,6546,6768,8789,58756,5645,5657,6556,6756,6756]
    
}
```

```python
import pandas as pd
```

```python
df=pd.DataFrame(data)
```

```python
df
```

```python
## Standardization
## z=x−μ/σ
```

```python
df["scalesalary"]= (df["Empsalary"]-df["Empsalary"].mean())/df["Empsalary"].std()
```

```python
df
```

```python
df.scalesalary.max()
```

```python
df=df[df.scalesalary<df.scalesalary.max()]
```

```python
df
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
newdf=StandardScaler().fit_transform(df)
```

```python
newdf
```

### Normalization

```python
## z=(x-min)/(max-min)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
df=MinMaxScaler().fit_transform(newdf)
```
